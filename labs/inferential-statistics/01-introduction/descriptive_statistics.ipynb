# Descriptive Statistics Lab

This notebook demonstrates how to compute and visualize descriptive statistics using Python, NumPy, Pandas, and SciPy.

```python
import numpy as np
import pandas as pd
from scipy import stats
import seaborn as sns
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Generate a right-skewed dataset
data = np.random.lognormal(mean=0, sigma=1, size=1000)
df = pd.DataFrame(data, columns=['Value'])
```

```python
# Calculate central tendency
mean_val = df['Value'].mean()
median_val = df['Value'].median()

print(f"Mean:   {mean_val:.4f}")
print(f"Median: {median_val:.4f}")
```

```python
# Dispersion
variance = df['Value'].var(ddof=1)
std_dev = df['Value'].std(ddof=1)
iqr = stats.iqr(df['Value'])

print(f"Variance: {variance:.4f}")
print(f"Std Dev:  {std_dev:.4f}")
print(f"IQR:      {iqr:.4f}")
```

```python
# Morphology (Shape)
skewness = df['Value'].skew()
kurtosis = df['Value'].kurtosis()

print(f"Skewness: {skewness:.4f}")
print(f"Kurtosis: {kurtosis:.4f}")
```

```python
# Plotting the distribution
plt.figure(figsize=(10, 5))
sns.histplot(df['Value'], kde=True, bins=50, color='blue')
plt.axvline(mean_val, color='red', linestyle='--', label=f'Mean: {mean_val:.2f}')
plt.axvline(median_val, color='green', linestyle='-', label=f'Median: {median_val:.2f}')
plt.title("Right-Skewed Distribution")
plt.legend()
plt.show()
```
