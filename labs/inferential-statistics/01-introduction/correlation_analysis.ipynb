### Key Topics Covered:

5. **Non-Linear Pitfall:** Demonstrating cases where correlation is zero despite deterministic relationships.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style="whitegrid")
np.random.seed(42)
```

## 1. Covariance Matrix & Pearson Correlation

We analyze the relationship between raw material imports and finished product exports.

```python
# Bivariate factory production sample
imports = pd.Series([10, 11, 14, 14, 21], name="Imports_Tons")
exports = pd.Series([12, 14, 15, 16, 23], name="Exports_Tons")

df = pd.DataFrame({"Imports_Tons": imports, "Exports_Tons": exports})

# Sample covariance (ddof=1)
cov_matrix = df.cov()
cov_val = cov_matrix.loc["Imports_Tons", "Exports_Tons"]

# Pearson correlation
corr_matrix = df.corr(method="pearson")
r_val, p_val = stats.pearsonr(df["Imports_Tons"], df["Exports_Tons"])

print("--- Covariance & Correlation Results ---")
print(f"Sample Covariance: {cov_val:.4f}")
print(f"Pearson r:         {r_val:.4f}")
print(f"p-value:           {p_val:.4e}")
```

## 2. Scale Invariance Experiment

Notice what happens when we convert tons to kilograms (multiplying by 1,000).

```python
df["Imports_Kg"] = df["Imports_Tons"] * 1000

cov_kg = df[["Imports_Kg", "Exports_Tons"]].cov().loc["Imports_Kg", "Exports_Tons"]
r_kg, _ = stats.pearsonr(df["Imports_Kg"], df["Exports_Tons"])

print(f"Covariance with Kg:  {cov_kg:.4f} (Multiplied by 1000)")
print(f"Correlation with Kg: {r_kg:.4f}   (Completely unchanged!)")
```

## 3. Multivariate Correlation Heatmap

```python
# Generating synthetic financial dataset
n = 100
gdp = np.random.normal(100, 10, n)
inflation = -0.6 * gdp + np.random.normal(50, 5, n)
interest_rate = 0.7 * inflation + np.random.normal(5, 2, n)
unemployment = np.random.uniform(3, 10, n)

macro_df = pd.DataFrame({
    "GDP": gdp,
    "Inflation": inflation,
    "InterestRate": interest_rate,
    "Unemployment": unemployment
})

plt.figure(figsize=(8, 6))
sns.heatmap(macro_df.corr(), annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1, linewidths=0.5)
plt.title("Macroeconomic Correlation Matrix Heatmap")
plt.show()
```

## 4. The Non-Linearity Trap

Demonstrating that $r \approx 0$ does not imply that two variables are independent.

```python
x_quad = np.linspace(-10, 10, 200)
y_quad = x_quad ** 2  # Exact quadratic dependency

r_quad, _ = stats.pearsonr(x_quad, y_quad)

plt.figure(figsize=(7, 4))
plt.scatter(x_quad, y_quad, color="darkred", alpha=0.7)
plt.title(f"Deterministic Non-Linear Dependency: Y = X^2 (Pearson r = {r_quad:.4f})")
plt.xlabel("X")
plt.ylabel("Y = X^2")
plt.show()

print(f"Calculated Pearson correlation: {r_quad:.4f}")
print("Warning: Pearson r is 0 because there is no linear trend, even though Y is 100% deterministic on X!")
```
