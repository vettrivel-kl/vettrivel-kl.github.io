# Python Variables, Identifiers & Memory Pointer Lab

In Python, variable names are identifiers that bind labels to objects in memory. This lab tests legal vs. illegal identifier syntax using Python's built-in `str.isidentifier()` and `keyword.iskeyword()`, examines the silent danger of **shadowing built-in functions**, and demonstrates object identity references using `id()`.

```python
import keyword

candidates = [
    "student_name", "_internal", "2fast", "user-name", "class",
    "list", "MAX_SPEED", "StudentProfile", "total$count", "__init__"
]

print(f"{'Identifier':<16} {'Legal?':<8} {'Keyword?':<10} {'Diagnostic'}")
print("-" * 58)
for name in candidates:
    is_kw = keyword.iskeyword(name)
    is_id = name.isidentifier()
    legal = is_id and not is_kw
    
    if is_kw:
        diag = "Reserved Python keyword!"
    elif not is_id:
        diag = "Invalid syntax (digits, hyphens, or symbols)"
    elif name in dir(__builtins__):
        diag = "WARNING: Shadows a Python built-in!"
    else:
        diag = "Valid Python identifier"
        
    print(f"{name:<16} {str(legal):<8} {str(is_kw):<10} {diag}")
```

## 1. The Catastrophe of Shadowing Built-ins

Python allows assigning to built-in names like `list`, `str`, or `sum` without throwing a syntax error. However, doing so destroys the built-in function for the remainder of the session!

```python
# Demonstrating built-in shadowing
numbers = [10, 20, 30]
print(f"Original sum(): {sum(numbers)}")

# Accidental variable assignment shadowing sum
sum = 100 + 200  # Legal assignment, but now sum is an int!
print(f"Variable sum: {sum}")

try:
    sum(numbers)  # Attempting to call the original function
except TypeError as e:
    print(f"CRASH: {e} ('sum' is now an integer, not a function!)")

# Restoring the built-in
del sum
print(f"Restored sum(): {sum(numbers)}")
```

## 2. Variables as Labels: Mutable Reference Aliasing

Verify that `b = a` creates an alias pointing to the exact same object in memory using `id()` and the `is` operator.

```python
list_a = [1, 2, 3]
list_b = list_a  # Both labels point to the same list on the heap

print(f"id(list_a): {hex(id(list_a))}")
print(f"id(list_b): {hex(id(list_b))}")
print(f"list_a is list_b: {list_a is list_b}")

list_b.append(99)
print(f"\nAfter list_b.append(99):")
print(f"list_a: {list_a} (Modified through list_b!)")
print(f"list_b: {list_b}")
```
