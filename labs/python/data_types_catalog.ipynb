# Python Data Types, Collections & Hashability Lab

Python organizes data into **scalar primitives** (`int`, `float`, `str`, `bool`) and **collection data structures** (`list`, `tuple`, `dict`, `set`). This lab benchmarks indexing, in-place mutability, the shallow immutability trap of tuples, dictionary key hashability, and set algebra.

```python
# Initializing all 8 core types
samples = [
    ("age",         28),
    ("height",      5.9),
    ("name",        "Vettri"),
    ("is_student",  True),
    ("fruits",      ["apple", "banana"]),
    ("coordinates", (10, 20)),
    ("student",     {"name": "Vettri", "age": 28}),
    ("colors",      {"red", "green"})
]

print(f"{'Variable':<14} {'Type':<8} {'Mutable?':<10} Value")
print("-" * 58)
for label, val in samples:
    mut = "Mutable" if isinstance(val, (list, dict, set)) else "Immutable"
    print(f"{label:<14} {type(val).__name__:<8} {mut:<10} {val}")
```

## 1. Mutability: In-Place Modification vs. Immutability Errors

Verify that lists support in-place modification (`append()`, index assignment), while tuples reject item assignment with a `TypeError`.

```python
my_list = [10, 20, 30]
print(f"List before: {my_list} (id: {hex(id(my_list))})")
my_list[0] = 99
print(f"List after:  {my_list} (id: {hex(id(my_list))}) -> Same memory address!")

my_tuple = (10, 20, 30)
try:
    my_tuple[0] = 99
except TypeError as e:
    print(f"Tuple assignment failed: {e}")
```

## 2. The Shallow Tuple Trap: Nested Mutable Elements

A tuple containing a mutable object (like a list) is only shallowly immutable: the outer tuple reference cannot change, but the inner list can be mutated in-place!

```python
nested_tup = ("user", [1, 2, 3])
print(f"Tuple initially: {nested_tup}")

# Mutating the inner list
nested_tup[1].append(999)
print(f"Tuple modified:  {nested_tup} (Inner list mutated in-place!)")
```

## 3. Dictionary Key Hashability Requirements

Dictionary keys must be **hashable** (immutable). Tuples are hashable; lists are unhashable.

```python
valid_dict = {}
valid_dict[(10, 20)] = "Coordinate Key Successful!"
print(f"Tuple Key: {valid_dict[(10, 20)]}")

try:
    valid_dict[[10, 20]] = "List Key"
except TypeError as e:
    print(f"List Key failed: {e}")
```

## 4. Set Uniqueness & The Boolean Collapse

Because `True == 1` and `False == 0`, adding booleans to sets containing integers merges duplicates.

```python
s = {1, 3.14, "Python", True}
print(f"Set contents: {s}")
print(f"Set length:   {len(s)} (True was absorbed by integer 1!)")
```
