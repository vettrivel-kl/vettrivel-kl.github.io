# Python Values, Types & Dynamic Typing Lab

In Python, every value carries its type with it. Variables are labels pointing to objects in memory rather than rigid typed storage boxes. This lab explores the four foundational primitive types (`int`, `float`, `str`, `bool`), dynamic typing reassignments, floating-point precision quirks, boolean integer inheritance, and type casting.

```python
# Inspecting primitive types
num_value = 100
dec_value = 15.75
text_value = "Python"
flag_value = True

for name, val in [
    ('num_value', num_value),
    ('dec_value', dec_value),
    ('text_value', text_value),
    ('flag_value', flag_value)
]:
    print(f"{name:<12} | Value: {str(val):<8} | type: {type(val)} | __name__: {type(val).__name__}")
```

## 1. Type Inspection: `type()` vs. `isinstance()`

Why `isinstance()` is preferred for conditional logic, especially with tuples.

```python
val = 42
print(f"isinstance(42, int):           {isinstance(val, int)}")
print(f"isinstance(42, float):         {isinstance(val, float)}")
print(f"isinstance(42, (int, float)):  {isinstance(val, (int, float))} (Tuple check!)")
```

## 2. The Boolean Surprise: `bool` is a Subclass of `int`

In Python, `bool` inherits directly from `int`. `True == 1` and `False == 0`.

```python
print(f"isinstance(True, int):  {isinstance(True, int)}")
print(f"issubclass(bool, int):   {issubclass(bool, int)}")
print(f"True + True:             {True + True} (Arithmetic works!)")
print(f"True * 50:               {True * 50}")

# Counting true conditions
scores = [85, 42, 91, 67, 38]
passed_count = sum(s >= 60 for s in scores)
print(f"Students passed (sum of booleans): {passed_count}")
```

## 3. Floating-Point Precision: The `0.1 + 0.2` Trap

Computers store floating-point numbers in binary base-2. Numbers like $0.1$ and $0.2$ cannot be represented with finite binary fractions.

```python
a = 0.1 + 0.2
print(f"0.1 + 0.2 = {a:.17f}")
print(f"0.1 + 0.2 == 0.3: {a == 0.3} (Dangerous comparison!)")
print(f"abs(a - 0.3) < 1e-9: {abs(a - 0.3) < 1e-9} (Safe comparison)")
```

## 4. Type Casting and ValueError Traps

Converting strings, floats, and integers, and safely handling decimal string parsing.

```python
print(f"int(15.75):         {int(15.75)} (Truncates, does NOT round!)")
print(f"round(15.75):       {round(15.75)} (Proper rounding)")

try:
    int("15.75")
except ValueError as e:
    print(f"int('15.75') FAILED: {e}")
    print(f"Fix via float first: int(float('15.75')) = {int(float('15.75'))}")
```
