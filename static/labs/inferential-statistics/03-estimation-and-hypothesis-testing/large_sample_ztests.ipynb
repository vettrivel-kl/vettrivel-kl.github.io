# Large Sample Z-Tests Lab

This lab demonstrates:
1. Conducting One-Sample Z-Tests (Left, Right, Two-Tailed)
2. Computing P-values and Critical Values
3. Executing Independent Two-Sample Z-Tests
4. Validating equal variance assumptions with Levene's test

```python
import numpy as np
from scipy import stats
from statsmodels.stats.weightstats import ztest

# 1. One-Sample Two-Tailed Test (Car Mileage Scenario)
# H0: mu = 25 vs H1: mu != 25
x_bar = 24.0
mu_0 = 25.0
sigma = 2.5
n = 45
alpha = 0.01

z_calc = (x_bar - mu_0) / (sigma / np.sqrt(n))
p_value_two_tailed = 2 * (1 - stats.norm.cdf(abs(z_calc)))
z_crit = stats.norm.ppf(1 - alpha / 2)

print(f"Z-Calculated: {z_calc:.4f}")
print(f"P-Value:      {p_value_two_tailed:.5f}")
print(f"Critical Z:   ±{z_crit:.4f}")
print(f"Decision:     {'Reject H0' if abs(z_calc) > z_crit else 'Fail to Reject H0'}")
```

```python
# 2. One-Sample One-Tailed Tests (Left-Tailed vs Right-Tailed)
# Right-Tailed Delivery Time Scenario (H0: mu <= 60 vs H1: mu > 60)
z_deliv = (75.0 - 60.0) / (30.0 / np.sqrt(45))
p_deliv = 1 - stats.norm.cdf(z_deliv)
print(f"Right-Tailed Delivery Z: {z_deliv:.4f}, P-Value: {p_deliv:.5f}")

# Left-Tailed PVC Pipe Scenario (H0: mu >= 13 vs H1: mu < 13)
z_pvc = (12.5 - 13.0) / (1.0 / np.sqrt(900))
p_pvc = stats.norm.cdf(z_pvc)
print(f"Left-Tailed PVC Z:       {z_pvc:.4f}, P-Value: {p_pvc:.5f}")
```

```python
# 3. Two-Sample Z-Test for Difference in Means (Hemoglobin Scenario)
# H0: mu1 = mu2 vs H1: mu1 != mu2
m1, s1, n1 = 13.0, 4.1, 160
m2, s2, n2 = 15.0, 3.5, 180
alpha = 0.01

se_diff = np.sqrt((s1**2 / n1) + (s2**2 / n2))
z_two_samp = (m1 - m2) / se_diff
p_two_samp = 2 * (1 - stats.norm.cdf(abs(z_two_samp)))
z_crit_01 = stats.norm.ppf(1 - alpha / 2)

print(f"Two-Sample Z: {z_two_samp:.4f}")
print(f"P-Value:      {p_two_samp:.6f}")
print(f"Critical Z:   ±{z_crit_01:.4f}")
print(f"Decision:     {'Reject H0' if abs(z_two_samp) > z_crit_01 else 'Fail to Reject H0'}")
```
