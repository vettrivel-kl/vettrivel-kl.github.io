# Estimation and Confidence Intervals Lab

This lab covers:
1. Point Estimation of population parameters (Mean, Proportion)
2. Margin of Error computation
3. Confidence Intervals for Means and Proportions using `scipy.stats`
4. Sample Size determination given a desired Margin of Error

```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

# 1. Point Estimation (Portfolio Scenario)
N_total = 50
n_sample = 13
x_underperforming = 8

p_hat = x_underperforming / n_sample
estimated_total = p_hat * N_total

print(f"Sample Proportion: {p_hat:.4f}")
print(f"Estimated Total Underperforming: {round(estimated_total)}")
```

```python
# 2. Confidence Interval for Population Mean (London Income Scenario)
n = 250
x_bar = 45000
sigma = 4000

# 95% Confidence Interval
ci_95 = stats.norm.interval(0.95, loc=x_bar, scale=sigma / np.sqrt(n))
ci_90 = stats.norm.interval(0.90, loc=x_bar, scale=sigma / np.sqrt(n))
ci_99 = stats.norm.interval(0.99, loc=x_bar, scale=sigma / np.sqrt(n))

print(f"90% CI: {ci_90}")
print(f"95% CI: {ci_95}")
print(f"99% CI: {ci_99}")

# Notice: Higher confidence level yields a wider confidence interval
```

```python
# 3. Margin of Error & Sample Size Determination
# Formula: n = (z_{alpha/2} * sigma / E)^2
desired_E = 2.46
sigma_val = 15.0
alpha = 0.10
z_crit = stats.norm.ppf(1 - alpha / 2)

required_n = (z_crit * sigma_val / desired_E) ** 2
print(f"Z-Critical (90%): {z_crit:.4f}")
print(f"Required Sample Size n: {np.ceil(required_n)}")
```
