```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Plotting style configuration
sns.set_theme(style="whitegrid")
np.random.seed(42)
```

## 1. Measures of Central Tendency

We simulate price data with an asymmetric distribution to compare how different averages behave.

```python
prices = pd.Series([10, 12, 14, 15, 15, 18, 20, 22, 25, 30, 85])

arithmetic_mean = prices.mean()
median_val = prices.median()
mode_val = prices.mode()[0]
trimmed_mean_10 = stats.trim_mean(prices, proportiontocut=0.10)
geometric_mean = stats.gmean(prices)
harmonic_mean = stats.hmean(prices)

print("--- Central Tendency Measures ---")
print(f"Arithmetic Mean: {arithmetic_mean:.2f}")
print(f"10% Trimmed Mean: {trimmed_mean_10:.2f}")
print(f"Geometric Mean:  {geometric_mean:.2f}")
print(f"Harmonic Mean:   {harmonic_mean:.2f}")
print(f"Median:          {median_val:.2f}")
print(f"Mode:            {mode_val:.2f}")
print(f"\nNotice the inequality: Harmonic Mean <= Geometric Mean <= Arithmetic Mean holds true!")
```

## 2. Measures of Dispersion & Tukey's Outlier Rule

Here we compute the five-number summary and apply the $1.5 \times \text{IQR}$ threshold rule to isolate outliers.

```python
q1 = prices.quantile(0.25)
q3 = prices.quantile(0.75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

sample_var = prices.var(ddof=1)       # Bessel's correction (n - 1)
sample_std = prices.std(ddof=1)
cv = stats.variation(prices, ddof=1) * 100  # Coefficient of Variation (%)

outliers = prices[(prices < lower_fence) | (prices > upper_fence)].tolist()

print("--- Dispersion Measures ---")
print(f"Range:                  {prices.max() - prices.min():.2f}")
print(f"Q1 (25th Percentile):   {q1:.2f}")
print(f"Q3 (75th Percentile):   {q3:.2f}")
print(f"IQR:                    {iqr:.2f}")
print(f"Tukey Lower Fence:      {lower_fence:.2f}")
print(f"Tukey Upper Fence:      {upper_fence:.2f}")
print(f"Detected Outliers:      {outliers}")
print(f"Sample Variance (s^2):  {sample_var:.2f}")
print(f"Sample Std Dev (s):     {sample_std:.2f}")
print(f"Coefficient of Var:     {cv:.2f}%")
```

## 3. Shape Analysis: Skewness & Kurtosis

We compute Bowley's quartile skewness, Karl Pearson's skewness, and Fisher's excess kurtosis.

```python
# Bowley's Skewness based on quartiles
bowley_skew = (q3 - 2 * median_val + q1) / (q3 - q1)

# Karl Pearson's Skewness
pearson_skew = (3 * (arithmetic_mean - median_val)) / sample_std

# Fisher-Pearson Standardized Moments
moment_skew = prices.skew()
excess_kurtosis = prices.kurt()

print("--- Shape Coefficients ---")
print(f"Bowley Skewness (Quartiles): {bowley_skew:.4f}")
print(f"Pearson Skewness (Mean/Med): {pearson_skew:.4f}")
print(f"Fisher-Pearson Skewness:     {moment_skew:.4f} (Positive -> Right-skewed)")
print(f"Excess Kurtosis:             {excess_kurtosis:.4f} (> 0 -> Leptokurtic, heavy tails)")
```

## 4. Visual Diagnostics

Visualizing the distribution using a Box Plot alongside a KDE Histogram.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot showing Five-Number Summary and Outlier Points
sns.boxplot(x=prices, ax=axes[0], color="skyblue")
axes[0].set_title("Box Plot (Five-Number Summary & Outliers)")
axes[0].set_xlabel("Price")

# Histogram & KDE showing asymmetry
sns.histplot(prices, kde=True, ax=axes[1], color="navy", bins=8)
axes[1].axvline(arithmetic_mean, color='red', linestyle='--', label=f'Mean: {arithmetic_mean:.1f}')
axes[1].axvline(median_val, color='green', linestyle='-', label=f'Median: {median_val:.1f}')
axes[1].set_title("Distribution Density (Right-Skewed Tail)")
axes[1].legend()

plt.tight_layout()
plt.show()
```
