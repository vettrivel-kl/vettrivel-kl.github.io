# Correlation and Covariance Lab

This lab demonstrates how to compute and visualize covariance and correlation structures.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set seed
np.random.seed(42)

# Generate continuous random variables
X = np.random.normal(50, 10, 500)
Y = 1.5 * X + np.random.normal(0, 15, 500)
Z = -2.0 * X + np.random.normal(0, 20, 500)
W = np.random.normal(50, 20, 500)

df = pd.DataFrame({'X': X, 'Y': Y, 'Z': Z, 'W': W})
```

```python
# Covariance Matrix
cov_mat = df.cov()
print("Covariance Matrix:")
print(cov_mat)
```

```python
# Pearson Correlation Matrix
corr_mat = df.corr(method='pearson')
print("\nPearson Correlation Matrix:")
print(corr_mat)
```

```python
# Visualizing the Correlation Matrix with a Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_mat, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Pairplot to visualize geometric relationships
sns.pairplot(df, corner=True, diag_kind='kde')
plt.suptitle('Bivariate Scatter Geometries', y=1.02)
plt.show()
```
