# Python Input Streams, Print Customization & CLI I/O Lab

Interactive applications rely on standard I/O streams: `sys.stdin` via `input()` and `sys.stdout` via `print()`. This lab explores why `input()` strictly returns a string, demonstrates safe type casting with exception handling, customizes `print(*objects, sep, end)`, and parses multi-value inputs using `split()` and `map()`.

```python
# 1. The input() String Rule
# Simulating runtime user input
simulated_text = "25"
print(f"Raw input return:  {simulated_text} (type: {type(simulated_text).__name__})")

# Must explicitly cast to int for arithmetic
age = int(simulated_text)
print(f"After int() cast:   {age} + 5 = {age + 5}")
```

## 2. Safe Parsing with Exception Handling

Handling invalid integer strings (e.g. text or decimals) without crashing the program.

```python
test_inputs = ["42", "3.14", "apple"]

for user_entry in test_inputs:
    try:
        val = int(user_entry)
        print(f"Input '{user_entry}': Successfully parsed as integer {val}")
    except ValueError:
        try:
            val_f = float(user_entry)
            print(f"Input '{user_entry}': Parsed as float {val_f} -> int: {int(val_f)}")
        except ValueError:
            print(f"Input '{user_entry}': FAILED: Non-numeric text!")
```

## 3. Controlling `print()`: `sep` and `end`

Customizing delimiters between values and line terminations.

```python
# Custom sep
print("Python", "Java", "C++", sep=", ")
print("2026", "09", "22", sep="-")
print("Header", "Value", sep=" | ")

# Custom end in loops
print("Countdown: ", end="")
for i in range(5, 0, -1):
    print(i, end="... ")
print("BLASTOFF!")
```

## 4. Multi-Value Input Parsing: `split()` and `map()`

The standard competitive programming pattern to parse multiple integers on a single line.

```python
line_input = "10 20 30"
tokens = line_input.split()
print(f"1. .split():        {tokens} (List of strings)")

a, b, c = map(int, tokens)
print(f"2. map(int, ...):   a={a}, b={b}, c={c} (Integers)")
print(f"3. Sum of inputs:   {a + b + c}")
```
