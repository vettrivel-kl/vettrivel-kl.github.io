# Python Expressions, Operators & Precedence Lab

Expressions combine values, variables, and operators to evaluate a result. This lab investigates arithmetic operators, the strict float return rule of division (`/`), negative floor division (`//`) rounding toward negative infinity, right-to-left exponentiation associativity (`**`), and PEMDAS order of operations.

```python
# The 5 fundamental arithmetic operations
operations = [
    ("15 + 25", 15 + 25, "Addition"),
    ("30 / 5",  30 / 5,  "True Division (Always Float)"),
    ("10 * 6",  10 * 6,  "Multiplication"),
    ("25 % 4",  25 % 4,  "Modulo (Remainder)"),
    ("2 ** 5",  2 ** 5,  "Exponentiation")
]

print(f"{'Expression':<12} {'Result':<10} {'Type':<10} {'Operation'}")
print("-" * 58)
for expr, res, op in operations:
    print(f"{expr:<12} {str(res):<10} {type(res).__name__:<10} {op}")
```

## 1. Division Nuances: `/` vs. `//` vs. `%`

Observe how standard division always returns float, while floor division rounds down toward $-\infty$.

```python
print(f"30 / 5:   {30 / 5}   (type: {type(30 / 5).__name__})")
print(f"23 // 5:  {23 // 5}  (Floor towards -inf: 4.6 -> 4)")
print(f"-23 // 5: {-23 // 5} (Floor towards -inf: -4.6 -> -5!)")
print(f"25 % 4:   {25 % 4}   (Remainder)")
print(f"-25 % 4:  {-25 % 4}  (Python modulo matches sign of divisor: 3)")
```

## 2. Right-to-Left Associativity of Exponentiation

In Python, `**` evaluates right-to-left, unlike all other binary arithmetic operators.

```python
val1 = 2 ** 3 ** 2
val2 = (2 ** 3) ** 2
val3 = 2 ** (3 ** 2)

print(f"2 ** 3 ** 2:    {val1} (Evaluates 3**2=9, then 2**9=512)")
print(f"(2 ** 3) ** 2:  {val2} (Evaluates 8**2=64)")
print(f"2 ** (3 ** 2):  {val3}")
```

## 3. String Operators: Concatenation vs. Sequence Repetition

Python overloads `+` and `*` for strings, but mixing them with integers causes `TypeError`.

```python
print(f"'5' + '5': {'5' + '5'} (String Concatenation)")
print(f"5 + 5:     {5 + 5} (Arithmetic Addition)")
print(f"'5' * 3:   {'5' * 3} (Sequence Repetition)")

try:
    print('5' + 5)
except TypeError as e:
    print(f"'5' + 5 FAILED: {e}")
```
