# Python F-Strings, Alignment & Format Specifiers Lab

Introduced in Python 3.6, formatted string literals (`f"..."`) provide an expressive, optimized, and readable syntax for string interpolation. This lab explores embedded expressions, numeric formatting (rounding, commas, percentages), field width alignment, literal brace escaping, and Python 3.8+ self-documenting debug expressions (`f"{x=}"`).

```python
name = "Vettri"
score = 91.5678
total = 1250000
ratio = 0.8742

# Basic and number formatting
print(f"Hello, {name}!")
print(f"Rounded marks:   {score:.2f} (Rounds, does not truncate!)")
print(f"With commas:     {total:,}")
print(f"Percentage:      {ratio:.1%}")
```

## 1. Field Width, Alignment & Fill Characters

Control layout using `<` (left), `^` (center), and `>` (right), with custom fill padding.

```python
text = "Python"
width = 16

print(f"|{text:<{width}}| (Left aligned)")
print(f"|{text:^{width}}| (Center aligned)")
print(f"|{text:>{width}}| (Right aligned)")
print(f"|{text:*^16}| (Custom asterisk fill)")
print(f"|{text:.>16}| (Dot leader fill)")
```

## 2. Table Construction with Dynamic Widths

Generate structured monospace tables by parameterizing field widths.

```python
students = [("Alice", 95.5), ("Bob", 82.0), ("Charlie", 91.25)]

print(f"{'Name':<15} {'Score':>10} {'Status':>10}")
print("-" * 37)
for s_name, s_score in students:
    status = "Distinction" if s_score >= 90 else "Pass"
    print(f"{s_name:<15} {s_score:>10.2f} {status:>10}")
```

## 3. Python 3.8+ Self-Documenting Debug Expressions (`=`) 

Adding an equals sign `=` prints the expression itself alongside its evaluated result.

```python
x = 42
y = 10

print(f"{x=}")
print(f"{y=}")
print(f"{x + y=}")
print(f"{x * y=}")
```

## 4. The Boolean Width Gotcha & Literal Braces

Because `bool` is an integer subclass, applying numeric width converts `True` to `1`.

```python
flag = True
print(f"Raw:              |{flag}|")
print(f"Numeric Width:    |{flag:<10}| (Formatted as integer 1!)")
print(f"Proper String:    |{str(flag):<10}| (Safe string conversion)")

# Printing literal braces
print(f"Escaped Braces:   {{Hello}}")
```
